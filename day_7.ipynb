```python
words = []
while True:
    line = input("Enter a word (blank line to finish): ").strip()
    if line == "":
        break
    if line not in words:
        words.append(line)
for word in words:
    print(word)
```

```python
pos = []
neg = []
zero = []
while True:
  val=input("enter an integer: ")
  if val=="":
    break
  val=int(val)
  if val<0:
    neg.append(val)
  elif val==0:
    zero.append(val)
  else:
    pos.append(val)
for n in neg:
  print(n)
for z in zero:
  print(z)
for p in pos:
  print(p)
```

# sets

```python
fruits=["apple","mango","apple","mango","banana"]
print(fruits)
```

```python
fruits_set={"apple","mango","apple","mango","banana"}
print(fruits_set)
```

```python
fruits_set2=set(fruits)
print(fruits_set2)
```

```python
fruits_set_upper={fruit.upper() for fruit in fruits}
print(fruits_set_upper)
```

```python
fruits_set.add("plum")
print(fruits_set)
```

```python
fruits_set.remove("mango")
print(fruits_set)
```

```python
fruits_set.discard("mango")
print(fruits_set)
#most preferred method
```
