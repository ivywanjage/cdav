```python
print("mean:",grade.mean())
```

```python
print(f"median:{grade.median()}")
```

```python
grade.mode()#no mode,it returns all the values
```

```python
#quartiles
print("Q1: ",grade.quantile(0.25))
print("Q2: ",grade.quantile(0.50))
print("Q3: ",grade.quantile(0.75))
```

q3--75% of the students scored less than or equal to 87.75

```python
#deciles
print("D1: ",grade.quantile(0.10))
print("D2: ",grade.quantile(0.20))
print("D3: ",grade.quantile(0.30))
```

```python
#percentile
print("P1: ",grade.quantile(0.30))
```

```python
max=grade.max()
print("highesr grade is ",max)
min=grade.min()
print("lowest grade is ",min)

range=max-min
print("range is ",range)

Q3=grade.quantile(0.75)
Q1=grade.quantile(0.25)
IQR=Q3-Q1
print("IQR is ",IQR)
```

```python
#variance/std deviation
var=grade.var()
std=grade.std()
var
```
