```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#figure=canvas,ax=plot/chart
fig,ax=plt.subplots()
ax.plot([1,2,3],[2,4,3])
plt.show()
```

```python
#pyplot api
plt.plot([1,2,3],[2,4,1])
plt.title("pyplot plot api")
plt.show()
```

```python
x=np.linspace(0,10,50)
fig, axes = plt.subplots(1,2,figsize=(8,5))
axes[0].plot(x,np.sin(x));axes[0].set_title("sin")
axes[1].plot(x,np.cos(x));axes[1].set_title("cos")
plt.tight_layout();plt.show()
```

```python
#bar chart
fig,ax=plt.subplots()
ax.bar(["A","B","C"],[5,7,3])
ax.set_title("bar chart")
plt.show()
```

```python
fig,ax=plt.subplots(1,3)
ax[0].bar(["A","B","C"],[5,7,3]);ax[0].set_title("bar chart")
ax[1].bar(["carrots","broccoli","tomato"],[4,7,5]);ax[1].set_title("veg chart")
ax[2].bar(["a","b","c"],[2,4,6]);ax[2].set_title("category chart")
plt.show()
```

```python
#scatter plot
rng=np.random.default_rng(0)
fig,ax=plt.subplots()
ax.scatter(rng.random(30),rng.random(30))
ax.set_title("scatter plot")
plt.show()
```

```python
#histogram
fig,ax=plt.subplots()
ax.hist(rng.normal(size=200),bins=15)#bins no.of bars
ax.set_title("histogram")
plt.show()
```

```python
#pie charts
fig,ax=plt.subplots()
ax.pie([40,35,25],
        labels=["A","B","C"],
        autopct="%1.0f%%")
ax.set_title("pie chart")
plt.show()
```

```python
rng=np.random.default_rng(42)
n=12
df=pd.DataFrame({
    "month":pd.date_range("2024-01-01",
                          periods=n,
                          freq="MS").strftime("%b"),
    "sales":rng.integers(50,100,n),
    "cost":rng.integers(20,60,n),
    "region":rng.choice(["East","West"],n),
})
df["profit"]=df["sales"]-df["cost"]#a new column
df.head()
```

```python
#simple line plot
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"])
plt.show()
```

how to tell lines apart

```python
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"],color="crimson")
ax.plot(df["month"],df["cost"],color="steelblue")
ax.plot(df["month"],df["profit"],color="seagreen")
plt.show()
```

```python
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"],linestyle="-",linewidth=2,marker="s",label="sales")
ax.plot(df["month"],df["cost"],linestyle="--",linewidth=2,marker="o",label="cost")
ax.plot(df["month"],df["profit"],linestyle=":",linewidth=2,marker="^",label="profit")
ax.legend()
ax.set_title("graph of sales,cost and profit")
ax.set_xlabel("months")
ax.set_ylabel("count")
plt.show()
```

```python
fig,ax=plt.subplots(1,2)
ax[0].barh(df["month"],df["sales"],color="pink")

ax[0].set_title("sales graph")
ax[0].set_xlabel("count")
ax[0].set_ylabel("months")

ax[1].barh(df["month"],df["cost"],color="beige")

ax[1].set_title("cost graph")
ax[1].set_xlabel("count")
ax[1].set_ylabel("months")

plt.show()
```

```python
x=np.arange(len(df))
fig,ax=plt.subplots()
ax.bar(x-0.2,df["sales"],0.4,label="sales",color="pink")
ax.bar(x+0.2,df["cost"],0.4,label="costs",color="beige")
ax.set_xticks(x,df["month"])
ax.set_xlabel("count")
ax.set_ylabel("months")
ax.set_title("graph of sales and costs",fontsize=14,fontweight="bold")
ax.legend()
plt.show()
```
