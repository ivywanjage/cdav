-values passed when calling the function between the brackets.they correspond to the arguments present when the function was created

```python
#addition function
def add_3(a,b):
  return a+b
```

```python
add_3(50,25)
```

```python
def greeter_3(lang):
  greetings={"fr":"bonjour",
           "sw":"jambo",
           "en-br":"hi",
           "jp":"konichiwa",
          "en-us":"howdy"}
  return greetings[lang]
```

```python
print(f"{greeter_3('sw')} John")
```

```python
def authenticator(password):
  defined_pass="secret"
  if password==defined_pass:
    return True
  else:
    return False
```

```python
passw=input("enter password ")
if authenticator(passw):
  print(True)
else:
  print(False)
```

```python
def sum_diff(x,y):
  s=x+y
  d=x-y
  return(s,d)
```

```python
sum_of_both,diff_of_both=sum_diff(100,50)
print(sum_of_both)
print(diff_of_both)
```

```python
year = int(input("Enter a year: "))
remainder = year % 12
if remainder == 8:
    animal = "Dragon"
elif remainder == 9:
    animal = "Snake"
elif remainder == 10:
    animal = "Horse"
elif remainder == 11:
    animal = "Sheep"
elif remainder == 0:
    animal = "Monkey"
elif remainder == 1:
    animal = "Rooster"
elif remainder == 2:
    animal = "Dog"
elif remainder == 3:
    animal = "Pig"
elif remainder == 4:
    animal = "Rat"
elif remainder == 5:
    animal = "Ox"
elif remainder == 6:
    animal = "Tiger"
elif remainder == 7:
    animal = "Hare"
print(f"{year} is the year of the {animal}.")
```
