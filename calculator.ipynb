```python
value_1=input("what is your first value?")
value_2=input("what is your second value?")
value_one=float(value_1)
value_two=float(value_2)
addition=value_one+value_two
subtraction=value_one-value_two
multiplication=value_one*value_two
division=value_one/value_two
integer_division=value_one//value_two
modulus=value_one%value_two
power=value_one**value_two

print("add:",addition)
print("minus:",subtraction)
print("times:",multiplication)
print("divide:",division)
print("floor:",integer_division)
print("remainder:",modulus)
print("power:",power)

```
