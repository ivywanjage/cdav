```python
n=int(input("enter your first positive integer "))
m=int(input("enter your second positive integer "))
d=min(m,n)
while m%d!=0 or n%d!=0:
  d=d-1
print(f"the greatest common divisor of {n} and {m} is {d}")
```

```python
message = input("Enter the message: ")
shift = int(input("Enter the shift amount (positive or negative): "))
shifted_message = ""
for char in message:
    if "a" <= char <= "z":
        pos = ord(char) - ord("a")
        new_pos = (pos + shift) % 26
        shifted_message += chr(new_pos + ord("a"))
    elif "A" <= char <= "Z":
        pos = ord(char) - ord("A")
        new_pos = (pos + shift) % 26
        shifted_message += chr(new_pos + ord("A"))
    else:
        shifted_message += char
print("Shifted message:", shifted_message)
```

```python
user_string = input("Enter a string to check if it's a palindrome: ")
start = 0
end = len(user_string) - 1
is_palindrome = True
while start < end:
    if user_string[start] != user_string[end]:
        is_palindrome = False
        break
    start += 1
    end -= 1
if is_palindrome:
    print(f'Success! "{user_string}" is a palindrome.')
else:
    print(f'Note: "{user_string}" is not a palindrome.')
```
