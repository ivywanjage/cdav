```python
import pandas as pd
```

```python
df=pd.read_csv("/content/kenya_transactions_clean.csv")
```

```python
df.pivot_table(
    index="county",
    columns="status",
    values="amount",
    aggfunc="sum",
)
```

a chart in one line

```python
import matplotlib.pyplot as plt
```

```python
sales=df.groupby("county")["amount"].sum().sort_values()
sales.plot(kind="barh")
plt.show()
```

```python
df["amount"].plot(kind="box")
```

```python
df.boxplot(column="amount",by="county")
```

```python
ax=sales.plot(kind="bar",color="#7C2A6C")
ax.set_title("Total sales by county,Jan-Jun 2026")
ax.set_ylabel("sales(kes)")
ax.set_xlabel("")
plt.xticks(rotation=90,ha="left")
plt.tight_layout()
```

# prettier charts-use seaborn

```python
import seaborn as sns
```

```python
#count of transactions per county,one line
sns.countplot(data=df,x="county")
```

```python
#average amount per county with error bars
sns.barplot(data=df,x="county",y="amount")
```

# saving a chart to a file

```python
ax=sales.plot(kind="bar")
ax.set_title("Total sales by county")
plt.savefig("sales_by_.county.png",dpi=150,bbox_inches="tight")
```

```python
df["amount"].plot(kind="hist",bins=30,title="sale sizes")
```

```python
df.groupby("county")["amount"].sum().sort_values().plot(kind="barh")
```

```python
MONTHS = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
df.groupby("month")["amount"].sum().reindex(MONTHS).plot(kind="line")
```
