```python
letter=input("enter your letter: ")
if letter=="a" or letter=="e" or letter=="i" or letter=="o" or letter=="u":
  print("your letter is a vowel")
elif letter=="y":
  print("sometimes your letter is a vowel")
else:
  print("your letter is a consonant")
```

```python
sound_level=int(input("enter the sound level of your noise in decibels"))
if sound_level>130:
  print("your sound level is greater than a jackhammer")
elif sound_level==130:
  print("your sound level is equal to a jackhammer")
elif sound_level<130 and sound_level>106:
  print("your sound level is greater than a gas lawnmower and less than a jackhammer")
elif sound_level==106:
  print("your sound level is equal to a gas lawnmower")
elif sound_level<106 and sound_level>70:
  print("your sound level is greater than an alarm clock and less than a gas lawnmower")
elif sound_level==70:
  print("your sound level is equal to an alarm clock")
elif sound_level<70 and sound_level>40:
  print("your sound level is greater than a quiet room and less than an alarm clock ")
elif sound_level==40:
  print("your sound level is equal to a quiet room")
else:
  print("your sound level is less than a quiet room")
```
