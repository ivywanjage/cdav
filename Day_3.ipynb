```python
write a program that takes a value from the user, the program should print
divisible by two if the value is divisible by two
divisible by three if the value is divisible by three
divisible by both two and three if the value is divisible by both two and three
other wise it should print not divisible by two and three
```

```python
user_value=float(input("enter a value "))
if (user_value)%2==0 and (user_value)%3==0:
  print("its divisible by both two and three")
elif (user_value)%2==0:
  print("its divisible by two")
elif (user_value)%3==0:
  print("its divisible by three")
else:
  print("its not divisible by both two and three")
```
